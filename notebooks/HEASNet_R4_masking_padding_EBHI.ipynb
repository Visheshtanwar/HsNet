```python
# =====================================================================================================
# HEASNet REVISION EXPERIMENT v2 (Reviewer 4, round 2) — ONE Kaggle cell, GPU on.
#
# WHAT CHANGED AGAINST v1 (fixes of errors found in the v1 outputs)
#   * A6 overlay masking: v1 used FIXED box coordinates, which did NOT cover the overlays (a teal strip of the
#     green box and the whole date/CVP/ID text stayed visible). v2 DETECTS the overlays in every image:
#       - green/teal endoscope-position box  -> colour + shape detection
#       - text and graphics in the black border (date, time, CVP, ID, Sex, D.O.B., NBI, "Comment", blocks)
#         -> every non-black pixel outside the circular/octagonal endoscopic field
#     Masked pixels are set to BLACK (the colour of the border), and the figure shows the EXACT tensor the model
#     receives. Controls hide a same-shape / same-area region of mucosa inside the field.
#     Detection rate of each overlay per class is reported (shows whether an overlay is class-correlated).
#   * Output files carry the dataset name (v1 Kvasir-V2 masking table was a copy of the HyperKvasir one).
#   * XAI figure: ALL classes (one row each), exact class names from the saved predictions file, full branch
#     names (SE / Window-attention / Height-axis attention); a CSV lists the image used in every row.
#     Faithfulness images are drawn at random (seed 42) instead of the first images of each class.
#   * A7 (new, inference only): effect of masking the zero-padded tokens in window attention (Minor comment 3).
#   * EBHI-SEG: transferred tensors split into parameters and buffers; specimens with images of both classes
#     counted correctly; specimen-cluster bootstrap stratified by specimen label (no double-counting).
#   * No base-model rerun and no retraining of any HyperKvasir / Kvasir-V2 model.
#
#  PART A  HyperKvasir + Kvasir-V2 (NO retraining)
#     A1 re-evaluation | A3 (off) missing ablation | A4 leakage-controlled accuracy | A5 XAI | A6 overlay masking
#     A7 padding-mask sensitivity
#  PART B  EBHI-SEG with the FINAL HEASNet (only training in this run, ~10 min)
#
#  ATTACH ON KAGGLE: (1) HyperKvasir dataset, (2) Kvasir-V2 dataset (same paths as your training run),
#  (3) EBHI-SEG dataset, (4) your results ZIP from the 7 Nov run uploaded as a Kaggle dataset.
#  !! Do NOT edit or redraw the saved figures afterwards (no image tools); use the PNGs exactly as saved.
# =====================================================================================================
import os, re, gc, sys, json, math, time, random, shutil, zipfile, warnings, datetime, copy
from pathlib import Path
from collections import defaultdict
import numpy as np, pandas as pd, matplotlib, matplotlib.pyplot as plt
from PIL import Image
from scipy import ndimage as ndi
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms as T
from sklearn.metrics import (accuracy_score, f1_score, precision_recall_fscore_support, confusion_matrix, roc_auc_score,
                             roc_curve, cohen_kappa_score, matthews_corrcoef)
from scipy.stats import binomtest, wilcoxon, fisher_exact
warnings.filterwarnings("ignore"); T0 = time.time()

# ======================================= SETTINGS (edit only here) =======================================
RESULTS   = "/kaggle/input/datasets/visheshtanwar26/results/HEASNet_FINAL_ALL_RESULTS"   # folder containing HyperKvasir/ and Kvasir_V2/
EBHI_ROOT = "/kaggle/input/datasets/orvile/ebhi-seg-colorectal-cancer/EBHI-SEG"
OUT       = "/kaggle/working/HEASNet_REVISION_RESULTS_v2"
PATH_REMAP = {}            # e.g. {"/kaggle/input/datasets/visheshtanwar26/processed-datatset": "/kaggle/input/processed-datatset"}

EBHI_CLASS_MAP = {"Normal": (0, "Non-carcinoma"), "Low-grade IN": (0, "Non-carcinoma"), "Adenocarcinoma": (1, "Carcinoma")}
EBHI_GROUP_BY_FOLDER = False   # False = group by specimen code only, so one specimen can never be in two splits

SEED, EPOCHS, WARMUP = 42, 40, 5
IMG, BATCH, WORKERS = 224, 16, 2
LR_HEAD, LR_BB, WD, SMOOTH, DROPOUT = 1e-4, 1e-5, 1e-5, 0.05, 0.30
EMBED, HEADS, WINDOW = 512, 8, 4

RUN_MISSING_ABLATION = False   # A3 stays OFF (no retraining); Comment 4 is answered by explaining the design
BOOT = 2000
FAITH_PER_CLASS, FAITH_STEPS = 10, 20
CORR_PER_CLASS = 150

# A6 overlay detection (works on a 448 x 448 copy of each image; check Fig_A6_*_detection_QC.png)
OV = dict(DARK_T=35,          # a pixel is "black border" if max(R,G,B) < DARK_T
          FIELD_OPEN=21,      # opening size (px) that removes text strokes but keeps the endoscopic field
          FIELD_GROW=3,       # px added around the field before border graphics are searched
          TEAL_R_MAX=80, TEAL_G_MIN=90, TEAL_GR_MIN=55, TEAL_GB_TOL=20,   # teal box colour: low R, high G, G >= B - tol
          GREEN_MIN_AREA=0.004, GREEN_MIN_FILL=0.55,                      # min area (fraction of image), min fill of bbox
          GREEN_MAX_X0=0.15, GREEN_MIN_CY=0.35,                           # box starts near the left edge, centre below 35% height
          TEXT_MIN=110, TEXT_SAT=60,                                      # text pixel: min(R,G,B) >= 110 and max-min <= 60 (white/grey)
          TEXT_NEIGH=15, TEXT_DARKFRAC=0.5,                               # ... with >= 50% black pixels within 15 px (text sits on black)
          TEXT_JOIN=6, MIN_SPECK=6,
          TEXT_LEFT_X0=0.12, TEXT_LINE_W=0.30,                            # a text line found at the left edge is extended to 30% of the width
          TEXT_RIGHT_X1=0.88, TEXT_RIGHT_W=0.30,                          # ... and one at the right edge to 30% from the right (overlay text runs over tissue)
          TEXT_BELOW=0.20, TEXT_TOPHAT=50,                                # 2nd pass: thin white strokes up to 20% of the height below a found text block
          MARGIN=4)                             # join characters into lines; drop specks; grow masks by 4 px (=2 px at 224)

DPI = 300
QUICK = os.environ.get("HEAS_QUICK") == "1"
if QUICK:
    RESULTS, EBHI_ROOT, OUT = os.environ["HEAS_RESULTS"], os.environ["HEAS_EBHI"], os.environ["HEAS_OUT"]
    EPOCHS, WARMUP, BOOT, FAITH_PER_CLASS, FAITH_STEPS, CORR_PER_CLASS, WORKERS, DPI = 2, 1, 50, 1, 4, 3, 0, 60
PRETRAINED = not QUICK

# names used in figures and tables (class names are taken from the saved predictions file, never typed by hand)
BRANCH_SHORT = {"se": "SE", "window_attention": "WIN", "axial_attention": "AX"}
BRANCH_LONG  = {"se": "SE branch (channel attention)", "window_attention": "Window-attention branch", "axial_attention": "Height-axis attention branch"}
def disp(c):  # display name: dataset folder name with the first letter capitalised, e.g. "bbps-0-1" -> "Bbps-0-1"
    c = str(c).replace("_", "-"); return c[:1].upper() + c[1:]
# =========================================================================================================

os.makedirs(OUT, exist_ok=True)
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu"); PIN = DEV.type == "cuda"
class Tee:
    def __init__(self, *s): self.s = s
    def write(self, x): [f.write(x) for f in self.s]
    def flush(self): [f.flush() for f in self.s]
_LOG = open(f"{OUT}/FULL_LOG.txt", "w", encoding="utf-8"); sys.stdout = Tee(sys.__stdout__, _LOG)
def banner(t, ch="="): print("\n" + ch * 100 + f"\n{t}\n" + ch * 100)
def seed_all(s=SEED):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if torch.cuda.is_available(): torch.cuda.manual_seed_all(s)
def save(fig, path): fig.savefig(path, dpi=DPI, bbox_inches="tight"); plt.show(); plt.close(fig)
def remap(p):
    for a, b in PATH_REMAP.items():
        if p.startswith(a): return b + p[len(a):]
    return p
plt.rcParams.update({"font.size": 9, "axes.spines.top": False, "axes.spines.right": False})
banner(f"HEASNet REVISION EXPERIMENT v2 | {datetime.datetime.now():%Y-%m-%d %H:%M} | {DEV} "
       f"{torch.cuda.get_device_name(0) if DEV.type == 'cuda' else ''} | torch {torch.__version__}")
PAPER = []
def paper(t=""): PAPER.append(t)

# ============================================ MODELS (identical to the main script) ============================================
class SqueezeExcitation(nn.Module):
    def __init__(self, c, r=16):
        super().__init__(); h = max(c // r, 8); self.fc1 = nn.Linear(c, h); self.fc2 = nn.Linear(h, c)
    def forward(self, x): return x * torch.sigmoid(self.fc2(F.relu(self.fc1(x.mean(dim=(2, 3))))))[:, :, None, None]
class WindowAttention(nn.Module):
    """As trained: zero padding 7x7 -> 8x8, 4x4 windows, padded tokens NOT masked (mask_pad=False).
    mask_pad=True is used only in A7 (inference-time sensitivity check); it adds no parameters."""
    def __init__(self, dim, heads=8, window=4):
        super().__init__(); self.window = window; self.attn = nn.MultiheadAttention(dim, heads, batch_first=True); self.norm = nn.LayerNorm(dim)
        self.mask_pad = False
    def forward(self, x):
        b, c, h, w_ = x.shape; w = self.window; ph, pw = (-h) % w, (-w_) % w
        if ph or pw: x = F.pad(x, (0, pw, 0, ph))
        hp, wp = h + ph, w_ + pw
        t = x.permute(0, 2, 3, 1).reshape(b, hp // w, w, wp // w, w, c).permute(0, 1, 3, 2, 4, 5).reshape(-1, w * w, c)
        kpm = None
        if self.mask_pad and (ph or pw):
            valid = torch.zeros(hp, wp, dtype=torch.bool, device=x.device); valid[:h, :w_] = True
            vm = valid.reshape(hp // w, w, wp // w, w).permute(0, 2, 1, 3).reshape(-1, w * w)       # (windows, tokens)
            kpm = (~vm)[None].expand(b, -1, -1).reshape(-1, w * w)                                  # True = ignore this key
            kpm = kpm & ~kpm.all(1, keepdim=True)                                                   # never mask a whole window
        t = self.norm(t + self.attn(t, t, t, key_padding_mask=kpm, need_weights=False)[0])
        x = t.reshape(b, hp // w, wp // w, w, w, c).permute(0, 1, 3, 2, 4, 5).reshape(b, hp, wp, c).permute(0, 3, 1, 2)
        return x[:, :, :h, :w_]
class AxialAttentionHeight(nn.Module):
    def __init__(self, dim, heads=8):
        super().__init__(); self.attn = nn.MultiheadAttention(dim, heads, batch_first=True); self.norm = nn.LayerNorm(dim)
    def forward(self, x):
        b, c, h, w = x.shape; t = x.permute(0, 3, 2, 1).reshape(b * w, h, c)
        t = self.norm(t + self.attn(t, t, t, need_weights=False)[0]); return t.reshape(b, w, h, c).permute(0, 3, 2, 1)
class HEASNet(nn.Module):
    def __init__(self, num_classes, pretrained=True, use_se=True, use_window=True, use_axial=True):
        super().__init__()
        self.backbone = models.efficientnet_v2_s(weights=models.EfficientNet_V2_S_Weights.DEFAULT if pretrained else None).features
        self.projection = nn.Conv2d(1280, EMBED, 1, bias=False)
        self.se = SqueezeExcitation(EMBED) if use_se else nn.Identity()
        self.window_attention = WindowAttention(EMBED, HEADS, WINDOW) if use_window else nn.Identity()
        self.axial_attention = AxialAttentionHeight(EMBED, HEADS) if use_axial else nn.Identity()
        self.dropout = nn.Dropout(DROPOUT); self.classifier = nn.Linear(EMBED, num_classes)
    def forward(self, x):
        x = self.axial_attention(self.window_attention(self.se(self.projection(self.backbone(x)))))
        return self.classifier(self.dropout(x.mean(dim=(2, 3))))
    def backbone_parameters(self): return list(self.backbone.parameters())
    def head_parameters(self): return [p for m in [self.projection, self.se, self.window_attention, self.axial_attention, self.classifier] for p in m.parameters()]

# ============================================ DATA / TRAINING / PREDICTION ============================================
MEAN, STD = [0.485, 0.456, 0.406], [0.229, 0.224, 0.225]
tf = T.Compose([T.Resize((IMG, IMG)), T.ToTensor(), T.Normalize(MEAN, STD)])
BLACK = ((torch.zeros(3) - torch.tensor(MEAN)) / torch.tensor(STD)).view(3, 1, 1)       # black pixel in normalised space
def pack(m): return np.packbits(m.reshape(-1).astype(bool))
def unpack(p): return np.unpackbits(p)[:IMG * IMG].reshape(IMG, IMG).astype(bool)
class ImgDS(Dataset):
    """masks: optional list of packed 224x224 boolean masks; masked pixels are set to black (as the image border)."""
    def __init__(self, paths, targets, masks=None): self.paths = list(paths); self.targets = [int(t) for t in targets]; self.masks = masks
    def __len__(self): return len(self.paths)
    def __getitem__(self, i):
        with Image.open(self.paths[i]) as im: x = tf(im.convert("RGB"))
        if self.masks is not None: x = torch.where(torch.from_numpy(unpack(self.masks[i]))[None], BLACK, x)
        return x, self.targets[i]
def loader(paths, targets, shuffle=False, seed=SEED, masks=None):
    g = torch.Generator().manual_seed(seed) if shuffle else None
    return DataLoader(ImgDS(paths, targets, masks), batch_size=BATCH, shuffle=shuffle, generator=g, num_workers=WORKERS, pin_memory=PIN, persistent_workers=WORKERS > 0)

@torch.no_grad()
def predict(model, ld):
    model.eval(); P = []
    for x, _ in ld:
        x = x.to(DEV, non_blocking=True)
        with torch.autocast(device_type=DEV.type, enabled=DEV.type == "cuda"): lo = model(x)
        P.append(torch.softmax(lo.float(), 1).cpu().numpy())
    return np.concatenate(P)

def train(model, tr_ld, va_ld, ckpt, tag):
    """Exactly the main-run protocol: 5 frozen-backbone epochs (head lr 1e-4) then 35 fine-tuning epochs
    (backbone 1e-5, head 1e-4), AdamW, label smoothing 0.05, AMP, checkpoint = best validation accuracy."""
    model.to(DEV); crit = nn.CrossEntropyLoss(label_smoothing=SMOOTH); scaler = torch.amp.GradScaler("cuda", enabled=DEV.type == "cuda")
    for p in model.backbone_parameters(): p.requires_grad = False
    head_opt = torch.optim.AdamW([{"params": model.head_parameters(), "lr": LR_HEAD}], weight_decay=WD)
    fine_opt = torch.optim.AdamW([{"params": model.backbone_parameters(), "lr": LR_BB}, {"params": model.head_parameters(), "lr": LR_HEAD}], weight_decay=WD)
    best, best_ep, hist = -1, 0, []
    for ep in range(1, EPOCHS + 1):
        if ep == WARMUP + 1:
            for p in model.backbone_parameters(): p.requires_grad = True
        opt = head_opt if ep <= WARMUP else fine_opt; model.train(); t0 = time.time(); tl = tc = tn = 0
        for x, y in tr_ld:
            x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True); opt.zero_grad(set_to_none=True)
            with torch.autocast(device_type=DEV.type, enabled=DEV.type == "cuda"): lo = model(x); loss = crit(lo, y)
            scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
            tl += loss.item() * len(y); tc += (lo.argmax(1) == y).sum().item(); tn += len(y)
        pv = predict(model, va_ld); yv = np.asarray(va_ld.dataset.targets); va = float((pv.argmax(1) == yv).mean())
        vl = float(F.nll_loss(torch.log(torch.from_numpy(pv).clamp_min(1e-8)), torch.from_numpy(yv)).item())
        hist.append({"epoch": ep, "train_loss": tl / tn, "train_acc": tc / tn, "val_loss": vl, "val_acc": va, "seconds": time.time() - t0})
        if va > best: best, best_ep = va, ep; torch.save(model.state_dict(), ckpt)
        print(f"[{tag}] epoch {ep:02d}/{EPOCHS} | train loss {tl/tn:.4f} acc {tc/tn:.4f} | val loss {vl:.4f} acc {va:.4f} | best {best:.4f} (ep {best_ep}) | {time.time()-t0:.0f}s")
    model.load_state_dict(torch.load(ckpt, map_location=DEV, weights_only=True))
    return model, pd.DataFrame(hist), best_ep, best

# ============================================ STATISTICS ============================================
def wilson(k, n):
    if n == 0: return (np.nan, np.nan)
    c = binomtest(int(k), int(n)).proportion_ci(method="wilson"); return (c.low, c.high)
def mcnemar(ok_a, ok_b):
    a = int((ok_a & ok_b).sum()); b = int((ok_a & ~ok_b).sum()); c = int((~ok_a & ok_b).sum()); d = int((~ok_a & ~ok_b).sum())
    return {"both_correct": a, "HEASNet_only": b, "other_only": c, "both_wrong": d, "discordant": b + c, "exact_p": float(binomtest(b, b + c, .5).pvalue) if b + c else 1.0}
def pct(x): return f"{100*x:.2f}%"
_trapz = getattr(np, "trapezoid", None) or np.trapz

# ============================================ EXPLAINABLE AI ============================================
def _mm(c): return (c - c.amin((1, 2), keepdim=True)) / (c.amax((1, 2), keepdim=True) - c.amin((1, 2), keepdim=True) + 1e-8)
def _up(m): return F.interpolate(m, size=(IMG, IMG), mode="bilinear", align_corners=False)[:, 0]
class XAI:
    """Grad-CAM on the last EfficientNetV2-S block (target class) + spatial maps of the SE / window / axial branches
    (channel-wise mean |activation| of each branch output, bilinear upsampling, min-max normalisation)."""
    def __init__(self, m):
        self.m, self.st, self.h, self.names = m, {}, [], []
        for p in m.parameters(): p.requires_grad_(True)
        self.h.append(m.backbone[-1].register_forward_hook(lambda mod, i, o: self.st.__setitem__("act", o)))
        for n in ["se", "window_attention", "axial_attention"]:
            if not isinstance(getattr(m, n), nn.Identity):
                self.names.append(n); self.h.append(getattr(m, n).register_forward_hook(lambda mod, i, o, k=n: self.st.__setitem__(k, o.detach())))
    def close(self): [h.remove() for h in self.h]
    def run(self, x, target):
        self.m.eval(); x = x.to(DEV)
        with torch.enable_grad():
            lo = self.m(x); tg = torch.as_tensor(target, device=DEV)
            g = torch.autograd.grad(lo.gather(1, tg[:, None]).sum(), self.st["act"])[0]
        cam = _mm(_up(F.relu((g.mean((2, 3), keepdim=True) * self.st["act"]).sum(1, keepdim=True)).detach()))
        return lo.detach(), cam.cpu().numpy(), {n: _mm(_up(self.st[n].abs().mean(1, keepdim=True))).cpu().numpy() for n in self.names}
def load_vis(p): return np.asarray(Image.open(p).convert("RGB").resize((IMG, IMG)))
def batch_of(paths): return torch.stack([tf(Image.open(p).convert("RGB")) for p in paths])
def tag(ds): return ds.replace("-", "").replace(" ", "")      # "Kvasir-V2" -> "KvasirV2" (used in every file name)

def xai_suite(model, paths, y, probs, classes, od, ds):
    banner(f"{ds}: EXPLAINABLE AI (Grad-CAM + attention-branch maps + faithfulness)", "-")
    xai = XAI(model); pred = probs.argmax(1); K = len(classes); paths = np.asarray(paths); dt = tag(ds); rng = np.random.default_rng(SEED)
    print("branches found in the model:", [BRANCH_LONG[n] for n in xai.names]); print("classes (from the predictions file):", [disp(c) for c in classes])
    # --- figure 1: ALL classes, the most confident correctly classified test image of each class
    ex, rows = [], []
    for c in range(K):
        idx = np.where((y == c) & (pred == c))[0]
        if not len(idx): print(f"  !! no correctly classified image for class {classes[c]}"); continue
        i = int(idx[np.argmax(probs[idx, c])]); ex.append(i)
        rows.append({"row": len(ex), "class": classes[c], "class_display": disp(classes[c]), "file": os.path.basename(paths[i]), "path": paths[i], "p_true_class": float(probs[i, c])})
    pd.DataFrame(rows).to_csv(f"{od}/{dt}_Fig_XAI_1_rows.csv", index=False)
    cols = ["Image", "Grad-CAM"] + [BRANCH_LONG[n] for n in xai.names]
    fig, ax = plt.subplots(len(ex), len(cols), figsize=(2.6 * len(cols), 2.6 * len(ex)), squeeze=False)
    for r, i in enumerate(ex):
        _, cam, mp = xai.run(batch_of([paths[i]]), [y[i]]); im = load_vis(paths[i])
        ax[r, 0].imshow(im); ax[r, 0].set_ylabel(disp(classes[y[i]]), fontsize=9, fontweight="bold")
        ax[r, 1].imshow(im); ax[r, 1].imshow(cam[0], cmap="jet", alpha=.45)
        for k, n in enumerate(xai.names): ax[r, 2 + k].imshow(im); ax[r, 2 + k].imshow(mp[n][0], cmap="jet", alpha=.45)
    for k, t in enumerate(cols): ax[0, k].set_title(t, fontsize=9)
    for a in ax.ravel(): a.set_xticks([]); a.set_yticks([])
    fig.suptitle(f"{ds}: Grad-CAM (true class) and attention-branch activation maps — {len(ex)} classes, most confident correct test image per class", y=1.0)
    fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_XAI_1_gradcam_and_attention_branches.png")
    # --- figure 2/3: Grad-CAM galleries (predicted class), correct and misclassified, 2 per class
    for correct, fn in [(True, "Fig_XAI_2_gradcam_correct.png"), (False, "Fig_XAI_3_gradcam_misclassified.png")]:
        sel = []
        for c in range(K):
            idx = np.where((y == c) & ((pred == c) if correct else (pred != c)))[0]; sel += idx[np.argsort(-probs[idx].max(1))][:2].tolist()
        if not sel: continue
        nc = 4; nr = math.ceil(len(sel) / nc); fig, ax = plt.subplots(nr, nc, figsize=(nc * 3.2, nr * 3.2), squeeze=False); [a.axis("off") for a in ax.ravel()]
        for k, i in enumerate(sel):
            _, cam, _ = xai.run(batch_of([paths[i]]), [pred[i]]); a = ax[k // nc, k % nc]; a.imshow(load_vis(paths[i])); a.imshow(cam[0], cmap="jet", alpha=.4)
            a.set_title(f"True: {disp(classes[y[i]])}\nPred: {disp(classes[pred[i]])} ({probs[i].max():.2f})", fontsize=7, color="green" if correct else "red")
        fig.suptitle(f"{ds}: Grad-CAM for the predicted class, {'correct' if correct else 'misclassified'} test images"); fig.tight_layout(); save(fig, f"{od}/{dt}_{fn}")
    # --- Pearson correlation Grad-CAM vs each branch map (per class)
    rows = []
    for c in range(K):
        idx = np.where(y == c)[0][:CORR_PER_CLASS]; v = {n: [] for n in xai.names}
        for s in range(0, len(idx), 16):
            ch = idx[s:s + 16]; _, cam, mp = xai.run(batch_of(paths[ch]), y[ch]); cf = cam.reshape(len(ch), -1); cf = (cf - cf.mean(1, keepdims=True)) / (cf.std(1, keepdims=True) + 1e-8)
            for n in xai.names: mf = mp[n].reshape(len(ch), -1); mf = (mf - mf.mean(1, keepdims=True)) / (mf.std(1, keepdims=True) + 1e-8); v[n] += (cf * mf).mean(1).tolist()
        rows.append({"class": disp(classes[c]), "n_images": len(idx), **{f"{BRANCH_SHORT[n]}_r_mean": float(np.mean(v[n])) for n in xai.names}, **{f"{BRANCH_SHORT[n]}_r_sd": float(np.std(v[n])) for n in xai.names}})
    cor = pd.DataFrame(rows); cor.to_csv(f"{od}/{dt}_gradcam_vs_branch_correlation.csv", index=False)
    print("\nPearson r between Grad-CAM and each attention-branch map (SE = channel attention, WIN = window attention, AX = height-axis attention)"); print(cor.round(3).to_string(index=False))
    cc = [f"{BRANCH_SHORT[n]}_r_mean" for n in xai.names]; v = cor[cc].values; lim = max(.3, np.nanmax(np.abs(v)))
    fig, ax = plt.subplots(figsize=(6, max(3, .45 * K))); im = ax.imshow(v, cmap="RdBu_r", vmin=-lim, vmax=lim, aspect="auto")
    ax.set_xticks(range(len(cc))); ax.set_xticklabels(["SE\n(channel)", "Window\nattention", "Height-axis\nattention"][:len(cc)]); ax.set_yticks(range(K)); ax.set_yticklabels(cor["class"])
    for i in range(K):
        for j in range(len(cc)): ax.text(j, i, f"{v[i,j]:.2f}", ha="center", va="center", fontsize=8)
    ax.set_title(f"{ds}: Pearson r, Grad-CAM vs branch map"); fig.colorbar(im, fraction=.05); fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_XAI_4_gradcam_branch_correlation.png")
    # --- deletion / insertion faithfulness vs random order (removed pixels -> 0 in normalised space = ImageNet mean)
    g = torch.Generator().manual_seed(SEED); fr = np.linspace(0, 1, FAITH_STEPS + 1); HW = IMG * IMG; rows = []; curves = defaultdict(list)
    rank = lambda s: torch.argsort(torch.argsort(-s)).view(-1)
    for c in range(K):
        idx = np.where((y == c) & (pred == c))[0]
        for i in sorted(rng.choice(idx, min(FAITH_PER_CLASS, len(idx)), replace=False)) if len(idx) else []:
            x = batch_of([paths[i]]); _, cam, _ = xai.run(x, [c]); x = x.to(DEV); out = {"class": disp(classes[c]), "path": paths[i]}
            for nm, rk in [("cam", rank(torch.from_numpy(cam[0]).flatten())), ("rand", rank(torch.rand(HW, generator=g)))]:
                for mode in ["del", "ins"]:
                    ims = [(x * ~m) if mode == "del" else (x * m) for m in [(rk < f * HW).reshape(1, 1, IMG, IMG).to(DEV) for f in fr]]
                    with torch.no_grad(): pr = torch.softmax(model(torch.cat(ims)).float(), 1)[:, c].cpu().numpy()
                    out[f"{mode}_{nm}_auc"] = float(_trapz(pr, fr)); curves[f"{mode}_{nm}"].append(pr)
            rows.append(out)
    xai.close(); fa = pd.DataFrame(rows); fa.to_csv(f"{od}/{dt}_faithfulness_per_image.csv", index=False)
    res = {"n_images": len(fa), "images_per_class": FAITH_PER_CLASS, "selection": "random correctly classified test images (seed 42)"}
    for mode, better in [("del", "lower"), ("ins", "higher")]:
        a, b = fa[f"{mode}_cam_auc"], fa[f"{mode}_rand_auc"]; win = (a < b) if mode == "del" else (a > b)
        p = float(wilcoxon(a, b).pvalue) if len(fa) > 1 and (a != b).any() else np.nan
        res[mode] = {"gradcam_auc_mean": float(a.mean()), "random_auc_mean": float(b.mean()), "better_is": better, "fraction_images_gradcam_better": float(win.mean()), "wilcoxon_p": p}
    json.dump(res, open(f"{od}/{dt}_faithfulness_summary.json", "w"), indent=2)
    print("\nFAITHFULNESS (deletion: lower AUC = more faithful; insertion: higher AUC = more faithful)"); print(json.dumps(res, indent=2))
    fig, axs = plt.subplots(1, 2, figsize=(10, 4))
    for ax, mode, ttl in [(axs[0], "del", "Deletion (most salient pixels removed first)"), (axs[1], "ins", "Insertion (most salient pixels revealed first)")]:
        for nm, col, lb in [("cam", "#1f6f8b", "Grad-CAM"), ("rand", "#888", "random order")]:
            a = np.array(curves[f"{mode}_{nm}"]); ax.plot(fr, a.mean(0), color=col, lw=2, label=f"{lb} (AUC {_trapz(a.mean(0), fr):.3f})"); ax.fill_between(fr, a.mean(0) - a.std(0), a.mean(0) + a.std(0), color=col, alpha=.15)
        ax.set(xlabel="Fraction of pixels", ylabel="Probability of true class", title=ttl); ax.legend()
    fig.suptitle(f"{ds}: Grad-CAM faithfulness, {len(fa)} correctly classified test images"); fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_XAI_5_faithfulness.png")
    return cor, res

# ============================================ A6 OVERLAY DETECTION AND MASKING ============================================
def _grow(m, px): return ndi.maximum_filter(m, size=2 * px + 1) if px > 0 else m
def _open(m, s): return ndi.maximum_filter(ndi.minimum_filter(m, size=s), size=s)
def _down(m): return m.reshape(IMG, 2, IMG, 2).any((1, 3))            # 448 -> 224, a 224-pixel is masked if any of its 4 pixels is
def detect_overlays(path):
    """Returns 224x224 boolean masks (green box, border text/graphics, endoscopic field) in the geometry of the model input."""
    S = 2 * IMG
    a = np.asarray(Image.open(path).convert("RGB").resize((S, S), Image.BILINEAR)).astype(np.int16); r, g, b = a[..., 0], a[..., 1], a[..., 2]
    # 1) green/teal endoscope-position box: colour + large, box-shaped component starting at the left edge
    teal = (r <= OV["TEAL_R_MAX"]) & (g >= OV["TEAL_G_MIN"]) & (g - r >= OV["TEAL_GR_MIN"]) & (g >= b - OV["TEAL_GB_TOL"])
    teal = _open(teal, 3); lab, n = ndi.label(teal); green = np.zeros((S, S), bool)
    for k, sl in enumerate(ndi.find_objects(lab), 1):
        if sl is None: continue
        area = int((lab[sl] == k).sum()); h, w = sl[0].stop - sl[0].start, sl[1].stop - sl[1].start
        if area >= OV["GREEN_MIN_AREA"] * S * S and area / (h * w) >= OV["GREEN_MIN_FILL"] and sl[1].start <= OV["GREEN_MAX_X0"] * S and (sl[0].start + sl[0].stop) / 2 >= OV["GREEN_MIN_CY"] * S:
            green[sl] = True                                               # whole box incl. the icon drawn inside it
    green = _grow(green, OV["MARGIN"])
    # 2) endoscopic field = largest bright region after an opening that removes thin text strokes
    nondark = a.max(-1) >= OV["DARK_T"]; core = _open(nondark & ~green, OV["FIELD_OPEN"]); lab, n = ndi.label(core)
    if n: field = ndi.binary_fill_holes(lab == (1 + int(np.argmax(ndi.sum(core, lab, range(1, n + 1))))))
    else: field = np.ones((S, S), bool)
    # 3) text / graphics = white or grey strokes that sit on a black background (date, time, CVP, ID, NBI, blocks, comments);
    #    specular highlights are rejected because they are surrounded by tissue, not by black pixels
    darkfrac = ndi.uniform_filter((a.max(-1) < OV["DARK_T"]).astype(np.float32), size=OV["TEXT_NEIGH"])
    cand = (a.min(-1) >= OV["TEXT_MIN"]) & (a.max(-1) - a.min(-1) <= OV["TEXT_SAT"]) & (darkfrac >= OV["TEXT_DARKFRAC"]) & ~green
    lab, n = ndi.label(_grow(cand, OV["TEXT_JOIN"])); text = np.zeros((S, S), bool)
    if n:
        cnt = ndi.sum(cand, lab, range(1, n + 1))
        for k, sl in enumerate(ndi.find_objects(lab), 1):
            if sl is None or cnt[k - 1] < OV["MIN_SPECK"]: continue
            y0, y1, x0, x1 = sl[0].start, sl[0].stop, sl[1].start, sl[1].stop           # bounding box of each text line / block
            if x0 <= OV["TEXT_LEFT_X0"] * S: x1 = max(x1, int(OV["TEXT_LINE_W"] * S))
            if x1 >= OV["TEXT_RIGHT_X1"] * S: x0 = min(x0, int((1 - OV["TEXT_RIGHT_W"]) * S))
            text[y0:y1, x0:x1] = True
    # 2nd pass (twice): lines of the same left text block that lie over tissue (e.g. "CVP", "G:N", white blocks)
    gray = a.mean(-1); tophat = gray - ndi.grey_opening(gray, size=9)
    thin = (a.min(-1) >= OV["TEXT_MIN"]) & (a.max(-1) - a.min(-1) <= OV["TEXT_SAT"]) & (tophat >= OV["TEXT_TOPHAT"]) & ~green
    for _ in range(2):
        left = text[:, :int(OV["TEXT_LINE_W"] * S)]
        if not left.any(): break
        rows = np.where(left.any(1))[0]; zone = np.zeros((S, S), bool)
        zone[rows.min():min(S, rows.max() + int(OV["TEXT_BELOW"] * S)), :int(OV["TEXT_LINE_W"] * S)] = True
        lab, n = ndi.label(_grow(thin & zone, OV["TEXT_JOIN"]))
        if not n: break
        cnt = ndi.sum(thin & zone, lab, range(1, n + 1))
        for k, sl in enumerate(ndi.find_objects(lab), 1):
            if sl is not None and cnt[k - 1] >= OV["MIN_SPECK"]: text[sl[0].start:sl[0].stop, :int(OV["TEXT_LINE_W"] * S)] = True
    text = _grow(text, OV["MARGIN"]) & ~green
    return _down(green), _down(text), _down(field)

def control_mask(mask, avoid, field, rng, tries=300):
    """Same-shape copy of `mask` moved to a random place inside the endoscopic field, not touching any overlay;
    if the shape does not fit, an equal-area square inside the field. Empty if nothing fits."""
    if not mask.any(): return mask.copy(), "none"
    ys, xs = np.where(mask); y0, y1, x0, x1 = ys.min(), ys.max() + 1, xs.min(), xs.max() + 1; shp = mask[y0:y1, x0:x1]; h, w = shp.shape
    for kind, piece in [("moved shape", shp), ("equal-area square", np.ones((min(IMG - 1, int(round(math.sqrt(mask.sum())))),) * 2, bool))]:
        h, w = piece.shape
        for _ in range(tries):
            ty, tx = rng.integers(0, IMG - h + 1), rng.integers(0, IMG - w + 1); new = np.zeros_like(mask); new[ty:ty + h, tx:tx + w] = piece
            if not (new & avoid).any() and (new & field).sum() >= .95 * new.sum(): return new, kind
    return np.zeros_like(mask), "did not fit"

def overlay_check(model, paths, y, base_probs, classes, od, ds):
    banner(f"{ds}: A6 OVERLAY MASKING / OCCLUSION CHECK (overlays detected per image; masked pixels set to black; no retraining)", "-")
    dt = tag(ds); N = len(paths); K = len(classes); rng = np.random.default_rng(SEED)
    G, Tx, Fd, CG, CT, info = [], [], [], [], [], []
    for i, p in enumerate(paths):
        g, t, f = detect_overlays(p); cg, kg = control_mask(g, g | t, f, rng); ct, kt = control_mask(t, g | t | cg, f, rng)
        G.append(g); Tx.append(t); Fd.append(f); CG.append(cg); CT.append(ct)
        info.append({"path": p, "class": classes[y[i]], "green_box": bool(g.any()), "green_area_%": 100 * g.mean(), "border_text": bool(t.any()), "text_area_%": 100 * t.mean(),
                     "control_green": kg, "control_text": kt})
        if (i + 1) % 500 == 0: print(f"  overlay detection {i+1}/{N}")
    info = pd.DataFrame(info); info.to_csv(f"{od}/{dt}_A6_overlay_detection_per_image.csv", index=False)
    conds = {"green box": G, "border text/graphics": Tx, "both overlays": [g | t for g, t in zip(G, Tx)],
             "control for green box (same shape moved into the field)": CG, "control for text (same area moved into the field)": CT}
    base = base_probs.argmax(1); bok = base == y; rows, per_class, preds = [], {}, {}
    for name, M in conds.items():
        has = np.array([m.any() for m in M]); pr = predict(model, loader(paths, y, masks=[pack(m) for m in M])).argmax(1); ok = pr == y; preds[name] = pr
        a = int((bok & ~ok).sum()); b = int((~bok & ok).sum()); cl = pd.Series(ok).groupby(y).mean() - pd.Series(bok).groupby(y).mean(); per_class[name] = cl
        rows.append({"masked_region": name, "images_with_region": int(has.sum()), "mean_masked_area_%": float(np.mean([m.mean() for m, h in zip(M, has) if h]) * 100) if has.any() else 0.0,
                     "accuracy_unmasked_%": 100 * bok.mean(), "accuracy_masked_%": 100 * ok.mean(), "change_pp_all": 100 * (ok.mean() - bok.mean()),
                     "change_pp_images_with_region": 100 * (ok[has].mean() - bok[has].mean()) if has.any() else 0.0,
                     "predictions_changed": int((pr != base).sum()), "correct_to_wrong": a, "wrong_to_correct": b,
                     "exact_McNemar_p": float(binomtest(a, a + b, .5).pvalue) if a + b else 1.0, "largest_class_drop": f"{disp(classes[int(cl.idxmin())])} ({100*cl.min():+.1f} pp)"})
    d = pd.DataFrame(rows); d.to_csv(f"{od}/{dt}_A6_overlay_masking_results.csv", index=False)
    print(f"\n{ds}: overlay masking (n = {N}; unmasked accuracy {100*bok.mean():.2f}%)"); print(d.round(4).to_string(index=False))
    pc = pd.DataFrame({"class": [disp(c) for c in classes], "n_test": np.bincount(y, minlength=K),
                       "green_box_present_%": [100 * info.green_box[y == c].mean() for c in range(K)], "border_text_present_%": [100 * info.border_text[y == c].mean() for c in range(K)],
                       "accuracy_unmasked_%": [100 * bok[y == c].mean() for c in range(K)],
                       **{f"change_pp_{k}": [100 * per_class[k].get(c, 0) for c in range(K)] for k in conds}})
    pc.to_csv(f"{od}/{dt}_A6_overlay_masking_per_class.csv", index=False); print(f"\n{ds}: overlay presence and accuracy change per class"); print(pc.round(1).to_string(index=False))
    # QC figure 1: one image per class (with an overlay if the class has one) — outlines on the input + the EXACT masked model inputs
    def model_input(p, m):
        x = tf(Image.open(p).convert("RGB")); x = torch.where(torch.from_numpy(m)[None], BLACK, x) if m is not None else x
        return (x * torch.tensor(STD).view(3, 1, 1) + torch.tensor(MEAN).view(3, 1, 1)).clamp(0, 1).permute(1, 2, 0).numpy()
    ex = []
    for c in range(K):
        idx = np.where(y == c)[0]; withov = idx[(info.green_box.values[idx]) | (info.border_text.values[idx])]; ex.append(int((withov if len(withov) else idx)[0]))
    cols = ["Input: green box (green) and text (red) detected", "Both overlays masked (model input)", "Control: green-box shape moved", "Control: text area moved"]
    fig, ax = plt.subplots(len(ex), 4, figsize=(11, 2.8 * len(ex)), squeeze=False)
    for r, i in enumerate(ex):
        ax[r, 0].imshow(model_input(paths[i], None))
        if G[i].any(): ax[r, 0].contour(G[i], levels=[.5], colors="lime", linewidths=1)
        if Tx[i].any(): ax[r, 0].contour(Tx[i], levels=[.5], colors="red", linewidths=.8)
        ax[r, 1].imshow(model_input(paths[i], G[i] | Tx[i])); ax[r, 2].imshow(model_input(paths[i], CG[i])); ax[r, 3].imshow(model_input(paths[i], CT[i]))
        ax[r, 0].set_ylabel(disp(classes[y[i]]), fontsize=9, fontweight="bold")
    for k, t in enumerate(cols): ax[0, k].set_title(t, fontsize=8)
    for a in ax.ravel(): a.set_xticks([]); a.set_yticks([])
    fig.suptitle(f"{ds}: overlay detection and masking (one test image per class; masked pixels are black)", y=1.0); fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_A6_overlay_masking_examples.png")
    # QC figure 2: 30 random test images with the detected regions (check detection quality before using the numbers)
    sel = rng.choice(N, min(30, N), replace=False); fig, ax = plt.subplots(5, 6, figsize=(15, 13)); [a.axis("off") for a in ax.ravel()]
    for k, i in enumerate(sel):
        a = ax.ravel()[k]; a.imshow(model_input(paths[i], None)); a.set_title(disp(classes[y[i]]), fontsize=7)
        if G[i].any(): a.contour(G[i], levels=[.5], colors="lime", linewidths=1)
        if Tx[i].any(): a.contour(Tx[i], levels=[.5], colors="red", linewidths=.8)
    fig.suptitle(f"{ds}: detection QC on 30 random test images (green = endoscope-position box, red = border text/graphics)"); fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_A6_detection_QC.png")
    # bar chart of the accuracy change per class
    fig, ax = plt.subplots(figsize=(max(6, .7 * K), 3.8)); xk = np.arange(K); wd = .38
    ax.bar(xk - wd / 2, pc["change_pp_both overlays"], wd, label="both overlays masked", color="#c0392b")
    ax.bar(xk + wd / 2, pc["change_pp_control for green box (same shape moved into the field)"], wd, label="control (green-box shape moved)", color="#7f8c8d")
    ax.axhline(0, c="k", lw=.8); ax.set_xticks(xk); ax.set_xticklabels(pc["class"], rotation=45, ha="right"); ax.set_ylabel("Change in class recall (pp)"); ax.legend(fontsize=8)
    ax.set_title(f"{ds}: effect of masking the overlays, per class"); fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_A6_per_class_change.png")
    return d, pc

# ============================================ A7 PADDING-MASK SENSITIVITY ============================================
def padding_check(model, paths, y, base_probs, od, ds):
    banner(f"{ds}: A7 WINDOW ATTENTION — effect of masking the zero-padded tokens (inference only, same weights)", "-")
    if isinstance(model.window_attention, nn.Identity): print("  no window attention in this model"); return None
    model.window_attention.mask_pad = True; pm = predict(model, loader(paths, y)); model.window_attention.mask_pad = False
    ok0, ok1 = base_probs.argmax(1) == y, pm.argmax(1) == y; mc = mcnemar(ok0, ok1)
    r = {"padding": "7x7 map zero-padded to 8x8 (15 of 64 positions); trained WITHOUT key masking",
         "accuracy_as_trained_%": 100 * ok0.mean(), "accuracy_padded_keys_masked_%": 100 * ok1.mean(),
         "predictions_changed": int((pm.argmax(1) != base_probs.argmax(1)).sum()), "mean_abs_prob_change": float(np.abs(pm - base_probs).max(1).mean()), **mc}
    json.dump(r, open(f"{od}/{tag(ds)}_A7_padding_mask_check.json", "w"), indent=2); print(json.dumps(r, indent=2)); return r

# =====================================================================================================
#                                     PART A: HyperKvasir + Kvasir-V2
# =====================================================================================================
SUMMARY = {}
def part_a(ds, sub):
    banner(f"PART A — {ds}", "#"); od = f"{OUT}/{sub}"; os.makedirs(od, exist_ok=True); R = f"{RESULTS}/{sub}"; dt = tag(ds)
    P = np.load(f"{R}/predictions_test.npz", allow_pickle=True)
    paths = [remap(str(p)) for p in P["paths"]]; y = P["y_true"].astype(int); classes = [str(c) for c in P["class_names"]]; K = len(classes)
    print(f"{ds}: {len(y)} test images, {K} classes: {', '.join(disp(c) for c in classes)}")
    missing = [p for p in paths if not os.path.exists(p)]
    if missing: raise FileNotFoundError(f"{len(missing)} test images not found (first: {missing[0]}). Attach the dataset or set PATH_REMAP.")
    # ---------- A1 re-evaluation ----------
    seed_all(); model = HEASNet(K, pretrained=False); model.load_state_dict(torch.load(f"{R}/best_HEASNet.pt", map_location="cpu", weights_only=True)); model.to(DEV).eval()
    probs = predict(model, loader(paths, y)); pred = probs.argmax(1); ok = pred == y
    same = int((pred == P["y_pred"]).sum())
    print(f"A1 RE-EVALUATION [{ds}]: {ok.sum()}/{len(y)} correct = {pct(ok.mean())} | saved run: {(P['y_pred'] == y).sum()} correct | identical predictions: {same}/{len(y)}")
    pr, rc, f1, sup = precision_recall_fscore_support(y, pred, labels=range(K), zero_division=0); lo, hi = wilson(ok.sum(), len(y))
    m = {"dataset": ds, "n_test": len(y), "correct": int(ok.sum()), "accuracy": float(ok.mean()), "accuracy_wilson_95CI": [lo, hi], "macro_precision": float(pr.mean()), "macro_recall": float(rc.mean()),
         "macro_f1": float(f1.mean()), "cohen_kappa": float(cohen_kappa_score(y, pred)), "mcc": float(matthews_corrcoef(y, pred)),
         "macro_auc_ovr": float(roc_auc_score(np.eye(K)[y], probs, average="macro")), "identical_to_saved_predictions": same}
    json.dump(m, open(f"{od}/{dt}_A1_heasnet_test_metrics.json", "w"), indent=2); print(json.dumps(m, indent=2))
    np.savez_compressed(f"{od}/{dt}_A1_heasnet_test_predictions_reevaluated.npz", paths=np.asarray(paths), y_true=y, probabilities=probs, class_names=np.asarray(classes))
    paper(f"\n### {ds}: HEASNet re-evaluated from saved checkpoint"); paper(f"  accuracy {pct(m['accuracy'])} ({m['correct']}/{len(y)}), Wilson 95% CI {pct(lo)}-{pct(hi)}; macro F1 {m['macro_f1']:.4f}; kappa {m['cohen_kappa']:.4f}; MCC {m['mcc']:.4f}; macro AUC {m['macro_auc_ovr']:.4f}")
    # ---------- A3 (switched off) ----------
    others = {}
    if ds == "HyperKvasir" and RUN_MISSING_ABLATION:
        man = pd.read_csv(f"{R}/exact_image_split_manifest.csv"); man["path"] = man["path"].map(remap)
        tr, va = man[man.split == "train"], man[man.split == "validation"]
        tp = man[man.split == "test"].set_index("path").loc[paths]; assert (tp.target.values == y).all(), "manifest test labels do not match predictions_test.npz"
        banner("A3: missing ablation '+ Window + Axial' (no SE)", "-")
        seed_all(); abl = HEASNet(K, PRETRAINED, use_se=False, use_window=True, use_axial=True)
        abl, h, be, ba = train(abl, loader(tr.path, tr.target, shuffle=True), loader(va.path, va.target), f"{od}/best_abl_Window_Axial_noSE.pt", "abl + Window + Axial")
        h.to_csv(f"{od}/history_abl_Window_Axial_noSE.csv", index=False); others["+ Window + Axial (new)"] = predict(abl, loader(paths, y))
        np.savez_compressed(f"{od}/test_predictions_abl_Window_Axial_noSE.npz", probs=others["+ Window + Axial (new)"], y=y)
        print(f"  best val acc {ba:.4f} at epoch {be}; test acc {pct((others['+ Window + Axial (new)'].argmax(1) == y).mean())}"); del abl; gc.collect(); torch.cuda.empty_cache() if torch.cuda.is_available() else None
    # ---------- A4 leakage-controlled accuracy ----------
    nd_file = f"{R}/near_duplicate_per_test_image.csv"
    if os.path.exists(nd_file):
        banner(f"A4: {ds} leakage-controlled accuracy of HEASNet (test images with NO near-duplicate in training)", "-")
        nd = pd.read_csv(nd_file); nd["path"] = nd["path"].map(remap); nd = nd.set_index("path").loc[paths]; dup = nd.near_duplicate.values.astype(bool); ham = nd.min_hamming_to_train.values
        rows = []
        for nm, pp in [("HEASNet (proposed)", probs)] + list(others.items()):
            o = pp.argmax(1) == y; r = {"model": nm, "all_%": 100 * o.mean(), "with_duplicate_%": 100 * o[dup].mean(), "no_duplicate_%": 100 * o[~dup].mean()}
            lo, hi = wilson(o[~dup].sum(), (~dup).sum()); r["no_duplicate_CI95"] = f"{100*lo:.2f}-{100*hi:.2f}"
            r["fisher_p_dup_vs_nodup"] = float(fisher_exact([[o[dup].sum(), (~o[dup]).sum()], [o[~dup].sum(), (~o[~dup]).sum()]])[1]); rows.append(r)
        leak = pd.DataFrame(rows); leak.to_csv(f"{od}/{dt}_A4_leakage_controlled_accuracy.csv", index=False)
        print(f"near-duplicate test images (dHash Hamming <= 4): {dup.sum()}/{len(y)}; hash distance 0 (identical 64-bit dHash, NOT verified pixel-identical): {(ham == 0).sum()}"); print(leak.round(3).to_string(index=False))
        th = pd.DataFrame([{"exclude_if_hamming_<=": t, "n_kept": int((ham > t).sum()), "HEASNet_accuracy_%": 100 * ok[ham > t].mean()} for t in [0, 2, 4, 6, 8]])
        th.to_csv(f"{od}/{dt}_A4_threshold_sensitivity.csv", index=False); print("\nThreshold sensitivity (HEASNet)"); print(th.round(2).to_string(index=False))
        paper(f"\n### {ds}: leakage-controlled accuracy ({(~dup).sum()} test images without a near-duplicate in training)"); paper(leak.round(3).to_string(index=False)); paper(th.round(2).to_string(index=False))
    # ---------- A7 padding, A5 XAI, A6 overlay masking ----------
    pad = padding_check(model, paths, y, probs, od, ds)
    cor, faith = xai_suite(model, paths, y, probs, classes, od, ds)
    ov, ovc = overlay_check(model, paths, y, probs, classes, od, ds)
    paper(f"\n### {ds}: A7 padding-mask sensitivity"); paper(json.dumps(pad, indent=2))
    paper(f"\n### {ds}: Grad-CAM faithfulness"); paper(json.dumps(faith, indent=2))
    paper(f"\n### {ds}: overlay masking (unmasked accuracy {pct(ok.mean())})"); paper(ov.round(4).to_string(index=False)); paper(ovc.round(1).to_string(index=False))
    SUMMARY[ds] = {"heasnet": m, "padding": pad, "faithfulness": faith, "overlay": ov.to_dict("records")}
    del model; gc.collect(); torch.cuda.empty_cache() if torch.cuda.is_available() else None

# =====================================================================================================
#                                     PART B: EBHI-SEG with the FINAL HEASNet
# =====================================================================================================
def part_b():
    banner("PART B — EBHI-SEG (final HEASNet, initialised from the 12-class HyperKvasir checkpoint)", "#"); od = f"{OUT}/EBHI_SEG"; os.makedirs(od, exist_ok=True)
    root = Path(EBHI_ROOT); assert root.exists(), f"EBHI_ROOT not found: {root}"
    EXT = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"}; recs = []
    for folder, (lab, _) in EBHI_CLASS_MAP.items():
        fd = root / folder
        if not fd.is_dir(): print("  skip, folder not found:", folder); continue
        imd = fd / "image" if (fd / "image").is_dir() else fd
        ims = [p for p in sorted(imd.rglob("*")) if p.suffix.lower() in EXT]
        for p in ims:
            sid = re.split(r"[-_]", p.stem, maxsplit=1)[0]
            recs.append({"path": str(p), "file": p.name, "source_folder": folder, "label": lab, "specimen_id": sid, "group": f"{folder}::{sid}" if EBHI_GROUP_BY_FOLDER else sid})
        print(f"  {folder:16s} -> label {lab}: {len(ims)} images")
    df = pd.DataFrame(recs); classes = [next(n for _, (l, n) in EBHI_CLASS_MAP.items() if l == L) for L in [0, 1]]
    # ---------- B1 grouping audit + specimen-level split ----------
    multi = df.groupby("specimen_id").source_folder.nunique(); multi = multi[multi > 1]
    mixed_lab = df.groupby("specimen_id").label.nunique(); mixed_lab = mixed_lab[mixed_lab > 1]
    print(f"\nB1 GROUPING: {df.specimen_id.nunique()} specimen codes; {len(multi)} codes occur in more than one class folder; "
          f"{len(mixed_lab)} codes have images of BOTH labels (non-carcinoma and carcinoma)" + (f" (e.g. {', '.join(multi.index[:5])})" if len(multi) else ""))
    print("   Grouping key:", "folder + specimen code" if EBHI_GROUP_BY_FOLDER else "specimen code only (all images of a specimen stay in one split)")
    print("   EBHI-SEG provides NO patient identifiers: this split is SPECIMEN-level, not verified patient-level.")
    gl = df.groupby("group").label.agg(lambda s: "mixed" if s.nunique() > 1 else str(s.iloc[0]))
    rnd = random.Random(SEED); split = {}
    for stratum in sorted(gl.unique()):
        g = sorted(gl[gl == stratum].index); rnd.shuffle(g); n = len(g); ntr = int(round(.70 * n)); nva = max(1, int(round(.15 * n))) if n >= 3 else 0
        for i, x in enumerate(g): split[x] = "train" if i < ntr else ("val" if i < ntr + nva else "test")
    df["split"] = df.group.map(split)
    ov = {f"{a}-{b}": len(set(df[df.split == a].specimen_id) & set(df[df.split == b].specimen_id)) for a, b in [("train", "val"), ("train", "test"), ("val", "test")]}
    print("   specimen codes shared between splits:", ov, "-> OK, no specimen leakage" if not any(ov.values()) else "-> !! LEAKAGE (set EBHI_GROUP_BY_FOLDER = False)")
    sp_lab = df.groupby("specimen_id").label.agg(lambda s: "mixed" if s.nunique() > 1 else ("carcinoma only" if s.iloc[0] == 1 else "non-carcinoma only"))
    tab = df.drop_duplicates("specimen_id").assign(kind=lambda d: d.specimen_id.map(sp_lab)).groupby(["split", "kind"]).size().unstack(fill_value=0)
    img = df.groupby(["split", "label"]).size().unstack(fill_value=0).rename(columns={0: "images_non-carcinoma", 1: "images_carcinoma"})
    split_tab = img.join(tab).assign(unique_specimens=lambda d: d[[c for c in tab.columns]].sum(1)); print(split_tab.to_string()); split_tab.to_csv(f"{od}/EBHI_split_summary.csv")
    df.to_csv(f"{od}/EBHI_split_manifest.csv", index=False)
    tr, va, te = [df[df.split == s].reset_index(drop=True) for s in ["train", "val", "test"]]
    tr_ld, va_ld, te_ld = loader(tr.path, tr.label, shuffle=True), loader(va.path, va.label), loader(te.path, te.label)
    y = te.label.values.astype(int); grp = te.specimen_id.values
    # ---------- B2 initialisation from the 12-class HyperKvasir checkpoint ----------
    seed_all(); model = HEASNet(2, pretrained=False); sd = torch.load(f"{RESULTS}/HyperKvasir/best_HEASNet.pt", map_location="cpu", weights_only=True)
    own = model.state_dict(); keep = {k: v for k, v in sd.items() if k in own and v.shape == own[k].shape}; model.load_state_dict({**own, **keep})
    skipped = [k for k in own if k not in keep]; pnames = {n for n, _ in model.named_parameters()}
    init = {"checkpoint": "HyperKvasir/best_HEASNet.pt (12-class final HEASNet, test accuracy 95.02%)",
            "state_dict_tensors_transferred": len(keep), "state_dict_tensors_total": len(own),
            "of_which_parameters_transferred": sum(k in pnames for k in keep), "parameters_total": len(pnames),
            "of_which_buffers_transferred (BatchNorm running mean/var/num_batches_tracked)": sum(k not in pnames for k in keep),
            "reinitialised": skipped}
    print("\nB2 INITIALISATION:", json.dumps(init, indent=2)); json.dump(init, open(f"{od}/EBHI_initialisation.json", "w"), indent=2)
    model, h, be, ba = train(model, tr_ld, va_ld, f"{od}/best_HEASNet_EBHI.pt", "HEASNet EBHI-SEG"); h.to_csv(f"{od}/history_HEASNet_EBHI.csv", index=False)
    probs = predict(model, te_ld); np.savez_compressed(f"{od}/EBHI_test_predictions_HEASNet.npz", paths=te.path.values, y_true=y, probabilities=probs, specimen=grp)
    fig, ax = plt.subplots(1, 2, figsize=(10, 3.8)); ax[0].plot(h.epoch, h.train_loss, label="train"); ax[0].plot(h.epoch, h.val_loss, label="validation"); ax[0].set_title("EBHI-SEG: loss")
    ax[1].plot(h.epoch, h.train_acc, label="train"); ax[1].plot(h.epoch, h.val_acc, label="validation"); ax[1].set_title(f"EBHI-SEG: accuracy (selected epoch {be})")
    for a in ax: a.axvline(WARMUP + .5, ls=":", c="gray"); a.set_xlabel("Epoch"); a.legend()
    fig.tight_layout(); save(fig, f"{od}/EBHI_Fig_B2_training_curves.png")
    # ---------- B4 metrics with uncertainty ----------
    banner("B4: EBHI-SEG test metrics with 95% CIs (image and specimen level)", "-"); rng = np.random.default_rng(SEED)
    sp_y = pd.Series(y).groupby(grp).max()                         # specimen label: carcinoma if ANY of its test images is carcinoma
    sp_kind = pd.Series(y).groupby(grp).agg(lambda s: "mixed" if s.nunique() > 1 else str(s.iloc[0]))
    strata = {k: sorted(sp_kind[sp_kind == k].index) for k in sp_kind.unique()}; where = {g: np.where(grp == g)[0] for g in set(grp)}
    def boot_auc(p1, cluster):
        out = []
        for _ in range(BOOT):
            if cluster: i = np.concatenate([where[g] for k, gs in strata.items() for g in rng.choice(gs, len(gs), replace=True)])
            else: i = rng.integers(0, len(y), len(y))
            if len(set(y[i])) == 2: out.append(roc_auc_score(y[i], p1[i]))
        return np.percentile(out, [2.5, 97.5]) if out else np.array([np.nan, np.nan])
    p1 = probs[:, 1]; yh = probs.argmax(1); tn, fp, fn, tp = confusion_matrix(y, yh, labels=[0, 1]).ravel(); auc = roc_auc_score(y, p1)
    sp = pd.DataFrame({"g": grp, "y": y, "p": p1}).groupby("g").agg(y=("y", "max"), p=("p", "mean"))
    sp_auc = roc_auc_score(sp.y, sp.p) if sp.y.nunique() == 2 else np.nan; sp_ok = ((sp.p >= .5).astype(int) == sp.y)
    ib, cb = boot_auc(p1, False), boot_auc(p1, True); a_ci, se_ci, spc_ci, s_ci = wilson(tp + tn, len(y)), wilson(tp, tp + fn), wilson(tn, tn + fp), wilson(sp_ok.sum(), len(sp))
    E = {"model": "HEASNet (proposed)", "test_images": len(y), "test_images_non-carcinoma": int((y == 0).sum()), "test_images_carcinoma": int((y == 1).sum()),
         "test_specimens_unique": int(len(sp)), "specimens_non-carcinoma_only": len(strata.get("0", [])), "specimens_carcinoma_only": len(strata.get("1", [])), "specimens_with_both_labels": len(strata.get("mixed", [])),
         "accuracy_%": 100 * (tp + tn) / len(y), "acc_CI95_wilson": f"{100*a_ci[0]:.2f}-{100*a_ci[1]:.2f}", "sensitivity_%": 100 * tp / (tp + fn), "sens_CI95": f"{100*se_ci[0]:.2f}-{100*se_ci[1]:.2f}",
         "specificity_%": 100 * tn / (tn + fp), "spec_CI95": f"{100*spc_ci[0]:.2f}-{100*spc_ci[1]:.2f}", "F1": f1_score(y, yh), "MCC": matthews_corrcoef(y, yh), "AUC": auc,
         "AUC_CI95_image_bootstrap": f"{ib[0]:.4f}-{ib[1]:.4f}", "AUC_CI95_specimen_cluster_bootstrap": f"{cb[0]:.4f}-{cb[1]:.4f}",
         "specimen_level_AUC (mean probability per specimen; specimen = carcinoma if any image is)": sp_auc, "specimen_level_accuracy_%": 100 * sp_ok.mean(), "spec_acc_CI95": f"{100*s_ci[0]:.2f}-{100*s_ci[1]:.2f}",
         "errors": f"{fp} FP / {fn} FN"}
    pd.Series(E).to_csv(f"{od}/EBHI_metrics_with_CI.csv", header=False); print(pd.Series(E).to_string())
    wrong = te[yh != y]; print("\nHEASNet misclassified test images:"); print(wrong[["file", "source_folder", "specimen_id"]].to_string(index=False) if len(wrong) else "  none")
    paper("\n### EBHI-SEG (final HEASNet, init from 12-class HyperKvasir checkpoint; same protocol; specimen-level split)")
    paper(f"  initialisation: {json.dumps(init)}"); paper(f"  shared specimen codes between splits: {ov}"); paper(split_tab.to_string()); paper(pd.Series(E).to_string())
    fig, ax = plt.subplots(1, 2, figsize=(10, 4)); cm = confusion_matrix(y, yh)
    ax[0].imshow(cm, cmap="Blues"); [ax[0].text(j, i, cm[i, j], ha="center", va="center", color="white" if cm[i, j] > cm.max() / 2 else "black") for i in range(2) for j in range(2)]
    ax[0].set_xticks([0, 1]); ax[0].set_yticks([0, 1]); ax[0].set_xticklabels(classes); ax[0].set_yticklabels(classes); ax[0].set_xlabel("Predicted"); ax[0].set_ylabel("True")
    ax[0].set_title(f"EBHI-SEG: confusion matrix ({len(y)} test images)")
    fpr, tpr, _ = roc_curve(y, p1); ax[1].plot(fpr, tpr, lw=2, label=f"HEASNet (AUC {auc:.4f})"); ax[1].plot([0, 1], [0, 1], "k:")
    ax[1].set(xlabel="False positive rate", ylabel="True positive rate", title="EBHI-SEG: ROC (test)"); ax[1].legend(fontsize=8, loc="lower right")
    fig.tight_layout(); save(fig, f"{od}/EBHI_Fig_B4_confusion_and_ROC.png")
    cor, faith = xai_suite(model, te.path.values, y, probs, classes, od, "EBHI-SEG")
    paper("\n### EBHI-SEG: Grad-CAM faithfulness"); paper(json.dumps(faith, indent=2))
    SUMMARY["EBHI-SEG"] = {"init": init, "metrics": E, "faithfulness": faith}
    del model; gc.collect(); torch.cuda.empty_cache() if torch.cuda.is_available() else None

# =====================================================================================================
for name, fn in [("HyperKvasir", lambda: part_a("HyperKvasir", "HyperKvasir")), ("Kvasir-V2", lambda: part_a("Kvasir-V2", "Kvasir_V2")), ("EBHI-SEG", part_b)]:
    try: fn()
    except Exception as e:
        import traceback; traceback.print_exc(); print(f"!! {name} FAILED: {e}")
open(f"{OUT}/REVISION_TABLES.txt", "w", encoding="utf-8").write("REVISION TABLES v2 (all numbers computed in this run)\n" + "=" * 80 + "\n" + "\n".join(PAPER))
json.dump(SUMMARY, open(f"{OUT}/REVISION_SUMMARY.json", "w"), indent=2, default=str)
banner(f"ALL DONE in {(time.time()-T0)/3600:.2f} h", "="); print("\n".join(PAPER)); _LOG.flush()
zp = OUT + ".zip"
with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
    for r, _, fs in os.walk(OUT):
        for f in fs:
            z.write(os.path.join(r, f), os.path.relpath(os.path.join(r, f), os.path.dirname(OUT)))
print(f"ZIP: {zp} ({os.path.getsize(zp)/1e6:.1f} MB)")
try:
    from IPython.display import FileLink, display; display(FileLink(os.path.relpath(zp)))
except Exception: pass
```

```python
# =====================================================================================================
# HEASNet A6b — OVERLAY MASKING WITH TWO FILLS (black and inpainted tissue) — ONE Kaggle cell, GPU on, ~15 min.
# No training. Uses the same saved checkpoints, the same overlay detection and the same control masks as v2.
# Question answered: when the green endoscope-position box is masked, does accuracy drop because the box is
# removed, or because a black bottom-left rectangle (present in many HyperKvasir *Polyps* images) is added?
#   fill "black"   = as in v2 (masked pixels black)
#   fill "inpaint" = masked pixels filled from the surrounding tissue (OpenCV Telea inpainting, radius 5)
# Also: (1) per-image predictions for every condition, (2) where changed predictions go (true -> new class),
#       (3) how often each class already contains a BLACK bottom-left box (pre-existing dataset artefact).
# Attach the same datasets as for v2 (HyperKvasir, Kvasir-V2, results of the 7 Nov run).
# =====================================================================================================
import os, re, gc, sys, json, math, time, random, warnings, zipfile, datetime
from collections import defaultdict
import numpy as np, pandas as pd, matplotlib.pyplot as plt, cv2
from PIL import Image
from scipy import ndimage as ndi
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms as T
from scipy.stats import binomtest
warnings.filterwarnings("ignore"); T0 = time.time()
# ======================================= SETTINGS (edit only here) =======================================
RESULTS   = "/kaggle/input/datasets/visheshtanwar26/results/HEASNet_FINAL_ALL_RESULTS"   # folder containing HyperKvasir/ and Kvasir_V2/
EBHI_ROOT = "/kaggle/input/datasets/orvile/ebhi-seg-colorectal-cancer/EBHI-SEG"
OUT       = "/kaggle/working/HEASNet_A6b_OVERLAY_FILL_CHECK"
PATH_REMAP = {}            # e.g. {"/kaggle/input/datasets/visheshtanwar26/processed-datatset": "/kaggle/input/processed-datatset"}

EBHI_CLASS_MAP = {"Normal": (0, "Non-carcinoma"), "Low-grade IN": (0, "Non-carcinoma"), "Adenocarcinoma": (1, "Carcinoma")}
EBHI_GROUP_BY_FOLDER = False   # False = group by specimen code only, so one specimen can never be in two splits

SEED, EPOCHS, WARMUP = 42, 40, 5
IMG, BATCH, WORKERS = 224, 16, 2
LR_HEAD, LR_BB, WD, SMOOTH, DROPOUT = 1e-4, 1e-5, 1e-5, 0.05, 0.30
EMBED, HEADS, WINDOW = 512, 8, 4

RUN_MISSING_ABLATION = False   # A3 stays OFF (no retraining); Comment 4 is answered by explaining the design
BOOT = 2000
FAITH_PER_CLASS, FAITH_STEPS = 10, 20
CORR_PER_CLASS = 150

# A6 overlay detection (works on a 448 x 448 copy of each image; check Fig_A6_*_detection_QC.png)
OV = dict(DARK_T=35,          # a pixel is "black border" if max(R,G,B) < DARK_T
          FIELD_OPEN=21,      # opening size (px) that removes text strokes but keeps the endoscopic field
          FIELD_GROW=3,       # px added around the field before border graphics are searched
          TEAL_R_MAX=80, TEAL_G_MIN=90, TEAL_GR_MIN=55, TEAL_GB_TOL=20,   # teal box colour: low R, high G, G >= B - tol
          GREEN_MIN_AREA=0.004, GREEN_MIN_FILL=0.55,                      # min area (fraction of image), min fill of bbox
          GREEN_MAX_X0=0.15, GREEN_MIN_CY=0.35,                           # box starts near the left edge, centre below 35% height
          TEXT_MIN=110, TEXT_SAT=60,                                      # text pixel: min(R,G,B) >= 110 and max-min <= 60 (white/grey)
          TEXT_NEIGH=15, TEXT_DARKFRAC=0.5,                               # ... with >= 50% black pixels within 15 px (text sits on black)
          TEXT_JOIN=6, MIN_SPECK=6,
          TEXT_LEFT_X0=0.12, TEXT_LINE_W=0.30,                            # a text line found at the left edge is extended to 30% of the width
          TEXT_RIGHT_X1=0.88, TEXT_RIGHT_W=0.30,                          # ... and one at the right edge to 30% from the right (overlay text runs over tissue)
          TEXT_BELOW=0.20, TEXT_TOPHAT=50,                                # 2nd pass: thin white strokes up to 20% of the height below a found text block
          MARGIN=4)                             # join characters into lines; drop specks; grow masks by 4 px (=2 px at 224)

DPI = 300
QUICK = os.environ.get("HEAS_QUICK") == "1"
if QUICK:
    RESULTS, EBHI_ROOT, OUT = os.environ["HEAS_RESULTS"], os.environ["HEAS_EBHI"], os.environ["HEAS_OUT"]
    EPOCHS, WARMUP, BOOT, FAITH_PER_CLASS, FAITH_STEPS, CORR_PER_CLASS, WORKERS, DPI = 2, 1, 50, 1, 4, 3, 0, 60
PRETRAINED = not QUICK

# names used in figures and tables (class names are taken from the saved predictions file, never typed by hand)
BRANCH_SHORT = {"se": "SE", "window_attention": "WIN", "axial_attention": "AX"}
BRANCH_LONG  = {"se": "SE branch (channel attention)", "window_attention": "Window-attention branch", "axial_attention": "Height-axis attention branch"}
def disp(c):  # display name: dataset folder name with the first letter capitalised, e.g. "bbps-0-1" -> "Bbps-0-1"
    c = str(c).replace("_", "-"); return c[:1].upper() + c[1:]
# =========================================================================================================

class SqueezeExcitation(nn.Module):
    def __init__(self, c, r=16):
        super().__init__(); h = max(c // r, 8); self.fc1 = nn.Linear(c, h); self.fc2 = nn.Linear(h, c)
    def forward(self, x): return x * torch.sigmoid(self.fc2(F.relu(self.fc1(x.mean(dim=(2, 3))))))[:, :, None, None]
class WindowAttention(nn.Module):
    """As trained: zero padding 7x7 -> 8x8, 4x4 windows, padded tokens NOT masked (mask_pad=False).
    mask_pad=True is used only in A7 (inference-time sensitivity check); it adds no parameters."""
    def __init__(self, dim, heads=8, window=4):
        super().__init__(); self.window = window; self.attn = nn.MultiheadAttention(dim, heads, batch_first=True); self.norm = nn.LayerNorm(dim)
        self.mask_pad = False
    def forward(self, x):
        b, c, h, w_ = x.shape; w = self.window; ph, pw = (-h) % w, (-w_) % w
        if ph or pw: x = F.pad(x, (0, pw, 0, ph))
        hp, wp = h + ph, w_ + pw
        t = x.permute(0, 2, 3, 1).reshape(b, hp // w, w, wp // w, w, c).permute(0, 1, 3, 2, 4, 5).reshape(-1, w * w, c)
        kpm = None
        if self.mask_pad and (ph or pw):
            valid = torch.zeros(hp, wp, dtype=torch.bool, device=x.device); valid[:h, :w_] = True
            vm = valid.reshape(hp // w, w, wp // w, w).permute(0, 2, 1, 3).reshape(-1, w * w)       # (windows, tokens)
            kpm = (~vm)[None].expand(b, -1, -1).reshape(-1, w * w)                                  # True = ignore this key
            kpm = kpm & ~kpm.all(1, keepdim=True)                                                   # never mask a whole window
        t = self.norm(t + self.attn(t, t, t, key_padding_mask=kpm, need_weights=False)[0])
        x = t.reshape(b, hp // w, wp // w, w, w, c).permute(0, 1, 3, 2, 4, 5).reshape(b, hp, wp, c).permute(0, 3, 1, 2)
        return x[:, :, :h, :w_]
class AxialAttentionHeight(nn.Module):
    def __init__(self, dim, heads=8):
        super().__init__(); self.attn = nn.MultiheadAttention(dim, heads, batch_first=True); self.norm = nn.LayerNorm(dim)
    def forward(self, x):
        b, c, h, w = x.shape; t = x.permute(0, 3, 2, 1).reshape(b * w, h, c)
        t = self.norm(t + self.attn(t, t, t, need_weights=False)[0]); return t.reshape(b, w, h, c).permute(0, 3, 2, 1)
class HEASNet(nn.Module):
    def __init__(self, num_classes, pretrained=True, use_se=True, use_window=True, use_axial=True):
        super().__init__()
        self.backbone = models.efficientnet_v2_s(weights=models.EfficientNet_V2_S_Weights.DEFAULT if pretrained else None).features
        self.projection = nn.Conv2d(1280, EMBED, 1, bias=False)
        self.se = SqueezeExcitation(EMBED) if use_se else nn.Identity()
        self.window_attention = WindowAttention(EMBED, HEADS, WINDOW) if use_window else nn.Identity()
        self.axial_attention = AxialAttentionHeight(EMBED, HEADS) if use_axial else nn.Identity()
        self.dropout = nn.Dropout(DROPOUT); self.classifier = nn.Linear(EMBED, num_classes)
    def forward(self, x):
        x = self.axial_attention(self.window_attention(self.se(self.projection(self.backbone(x)))))
        return self.classifier(self.dropout(x.mean(dim=(2, 3))))
    def backbone_parameters(self): return list(self.backbone.parameters())
    def head_parameters(self): return [p for m in [self.projection, self.se, self.window_attention, self.axial_attention, self.classifier] for p in m.parameters()]

def _grow(m, px): return ndi.maximum_filter(m, size=2 * px + 1) if px > 0 else m
def _open(m, s): return ndi.maximum_filter(ndi.minimum_filter(m, size=s), size=s)
def _down(m): return m.reshape(IMG, 2, IMG, 2).any((1, 3))            # 448 -> 224, a 224-pixel is masked if any of its 4 pixels is
def detect_overlays(path):
    """Returns 224x224 boolean masks (green box, border text/graphics, endoscopic field) in the geometry of the model input."""
    S = 2 * IMG
    a = np.asarray(Image.open(path).convert("RGB").resize((S, S), Image.BILINEAR)).astype(np.int16); r, g, b = a[..., 0], a[..., 1], a[..., 2]
    # 1) green/teal endoscope-position box: colour + large, box-shaped component starting at the left edge
    teal = (r <= OV["TEAL_R_MAX"]) & (g >= OV["TEAL_G_MIN"]) & (g - r >= OV["TEAL_GR_MIN"]) & (g >= b - OV["TEAL_GB_TOL"])
    teal = _open(teal, 3); lab, n = ndi.label(teal); green = np.zeros((S, S), bool)
    for k, sl in enumerate(ndi.find_objects(lab), 1):
        if sl is None: continue
        area = int((lab[sl] == k).sum()); h, w = sl[0].stop - sl[0].start, sl[1].stop - sl[1].start
        if area >= OV["GREEN_MIN_AREA"] * S * S and area / (h * w) >= OV["GREEN_MIN_FILL"] and sl[1].start <= OV["GREEN_MAX_X0"] * S and (sl[0].start + sl[0].stop) / 2 >= OV["GREEN_MIN_CY"] * S:
            green[sl] = True                                               # whole box incl. the icon drawn inside it
    green = _grow(green, OV["MARGIN"])
    # 2) endoscopic field = largest bright region after an opening that removes thin text strokes
    nondark = a.max(-1) >= OV["DARK_T"]; core = _open(nondark & ~green, OV["FIELD_OPEN"]); lab, n = ndi.label(core)
    if n: field = ndi.binary_fill_holes(lab == (1 + int(np.argmax(ndi.sum(core, lab, range(1, n + 1))))))
    else: field = np.ones((S, S), bool)
    # 3) text / graphics = white or grey strokes that sit on a black background (date, time, CVP, ID, NBI, blocks, comments);
    #    specular highlights are rejected because they are surrounded by tissue, not by black pixels
    darkfrac = ndi.uniform_filter((a.max(-1) < OV["DARK_T"]).astype(np.float32), size=OV["TEXT_NEIGH"])
    cand = (a.min(-1) >= OV["TEXT_MIN"]) & (a.max(-1) - a.min(-1) <= OV["TEXT_SAT"]) & (darkfrac >= OV["TEXT_DARKFRAC"]) & ~green
    lab, n = ndi.label(_grow(cand, OV["TEXT_JOIN"])); text = np.zeros((S, S), bool)
    if n:
        cnt = ndi.sum(cand, lab, range(1, n + 1))
        for k, sl in enumerate(ndi.find_objects(lab), 1):
            if sl is None or cnt[k - 1] < OV["MIN_SPECK"]: continue
            y0, y1, x0, x1 = sl[0].start, sl[0].stop, sl[1].start, sl[1].stop           # bounding box of each text line / block
            if x0 <= OV["TEXT_LEFT_X0"] * S: x1 = max(x1, int(OV["TEXT_LINE_W"] * S))
            if x1 >= OV["TEXT_RIGHT_X1"] * S: x0 = min(x0, int((1 - OV["TEXT_RIGHT_W"]) * S))
            text[y0:y1, x0:x1] = True
    # 2nd pass (twice): lines of the same left text block that lie over tissue (e.g. "CVP", "G:N", white blocks)
    gray = a.mean(-1); tophat = gray - ndi.grey_opening(gray, size=9)
    thin = (a.min(-1) >= OV["TEXT_MIN"]) & (a.max(-1) - a.min(-1) <= OV["TEXT_SAT"]) & (tophat >= OV["TEXT_TOPHAT"]) & ~green
    for _ in range(2):
        left = text[:, :int(OV["TEXT_LINE_W"] * S)]
        if not left.any(): break
        rows = np.where(left.any(1))[0]; zone = np.zeros((S, S), bool)
        zone[rows.min():min(S, rows.max() + int(OV["TEXT_BELOW"] * S)), :int(OV["TEXT_LINE_W"] * S)] = True
        lab, n = ndi.label(_grow(thin & zone, OV["TEXT_JOIN"]))
        if not n: break
        cnt = ndi.sum(thin & zone, lab, range(1, n + 1))
        for k, sl in enumerate(ndi.find_objects(lab), 1):
            if sl is not None and cnt[k - 1] >= OV["MIN_SPECK"]: text[sl[0].start:sl[0].stop, :int(OV["TEXT_LINE_W"] * S)] = True
    text = _grow(text, OV["MARGIN"]) & ~green
    return _down(green), _down(text), _down(field)

def control_mask(mask, avoid, field, rng, tries=300):
    """Same-shape copy of `mask` moved to a random place inside the endoscopic field, not touching any overlay;
    if the shape does not fit, an equal-area square inside the field. Empty if nothing fits."""
    if not mask.any(): return mask.copy(), "none"
    ys, xs = np.where(mask); y0, y1, x0, x1 = ys.min(), ys.max() + 1, xs.min(), xs.max() + 1; shp = mask[y0:y1, x0:x1]; h, w = shp.shape
    for kind, piece in [("moved shape", shp), ("equal-area square", np.ones((min(IMG - 1, int(round(math.sqrt(mask.sum())))),) * 2, bool))]:
        h, w = piece.shape
        for _ in range(tries):
            ty, tx = rng.integers(0, IMG - h + 1), rng.integers(0, IMG - w + 1); new = np.zeros_like(mask); new[ty:ty + h, tx:tx + w] = piece
            if not (new & avoid).any() and (new & field).sum() >= .95 * new.sum(): return new, kind
    return np.zeros_like(mask), "did not fit"


os.makedirs(OUT, exist_ok=True)
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu"); PIN = DEV.type == "cuda"
class Tee:
    def __init__(self, *s): self.s = s
    def write(self, x): [f.write(x) for f in self.s]
    def flush(self): [f.flush() for f in self.s]
_LOG = open(f"{OUT}/FULL_LOG_A6b.txt", "w", encoding="utf-8"); sys.stdout = Tee(sys.__stdout__, _LOG)
def banner(t, ch="="): print("\n" + ch * 100 + f"\n{t}\n" + ch * 100)
def remap(p):
    for a, b in PATH_REMAP.items():
        if p.startswith(a): return b + p[len(a):]
    return p
def save(fig, path): fig.savefig(path, dpi=DPI, bbox_inches="tight"); plt.show(); plt.close(fig)
def tag(ds): return ds.replace("-", "").replace(" ", "")
banner(f"HEASNet A6b overlay fill check | {datetime.datetime.now():%Y-%m-%d %H:%M} | {DEV}")

MEAN, STD = [0.485, 0.456, 0.406], [0.229, 0.224, 0.225]
resize = T.Resize((IMG, IMG)); to_t = T.Compose([T.ToTensor(), T.Normalize(MEAN, STD)])
def pack(m): return np.packbits(m.reshape(-1).astype(bool))
def unpack(p): return np.unpackbits(p)[:IMG * IMG].reshape(IMG, IMG).astype(bool)
def make_input(path, mask=None, fill="black"):
    """Exactly the training transform (Resize 224 -> ToTensor -> Normalize); masked pixels black or inpainted at 224 px."""
    with Image.open(path) as im: a = np.asarray(resize(im.convert("RGB"))).copy()
    if mask is not None and mask.any():
        if fill == "black": a[mask] = 0
        else: a = cv2.inpaint(a, mask.astype(np.uint8) * 255, 5, cv2.INPAINT_TELEA)
    return to_t(Image.fromarray(a)), a
class MDS(Dataset):
    def __init__(self, paths, masks, fill): self.p, self.m, self.f = list(paths), masks, fill
    def __len__(self): return len(self.p)
    def __getitem__(self, i): return make_input(self.p[i], None if self.m is None else unpack(self.m[i]), self.f)[0]
@torch.no_grad()
def predict(model, paths, masks=None, fill="black"):
    ld = DataLoader(MDS(paths, masks, fill), batch_size=BATCH, num_workers=WORKERS, pin_memory=PIN); P = []
    for x in ld:
        with torch.autocast(device_type=DEV.type, enabled=DEV.type == "cuda"): P.append(torch.softmax(model(x.to(DEV)).float(), 1).cpu().numpy())
    return np.concatenate(P)

def black_corner_box(path):
    """Pre-existing BLACK inset box in the bottom-left corner (same place as the green box): a dark, box-shaped area
    whose right and top edges border bright tissue."""
    S = 2 * IMG; a = np.asarray(Image.open(path).convert("RGB").resize((S, S), Image.BILINEAR)).astype(np.int16); dark = a.max(-1) < OV["DARK_T"]
    lab, n = ndi.label(_open(dark, 5))
    for k, sl in enumerate(ndi.find_objects(lab), 1):
        if sl is None: continue
        y0, y1, x0, x1 = sl[0].start, sl[0].stop, sl[1].start, sl[1].stop; h, w = y1 - y0, x1 - x0; comp = lab[sl] == k
        if x0 > .02 * S or y1 < .97 * S or h < .15 * S or w < .15 * S or w > .45 * S or h > .5 * S or comp.mean() < .85: continue
        right = a[y0 + h // 4: y1 - 2, min(S - 1, x1 + 3): min(S, x1 + 12)].max(-1); top = a[max(0, y0 - 12): max(1, y0 - 3), x0 + w // 4: x1 - 2].max(-1)
        if right.size and top.size and (right >= OV["DARK_T"]).mean() > .7 and (top >= OV["DARK_T"]).mean() > .7: return True
    return False

def run(ds, sub):
    banner(f"{ds}: A6b overlay masking with black and inpainted fill", "#"); od = f"{OUT}/{sub}"; os.makedirs(od, exist_ok=True); R = f"{RESULTS}/{sub}"; dt = tag(ds)
    P = np.load(f"{R}/predictions_test.npz", allow_pickle=True); paths = [remap(str(p)) for p in P["paths"]]; y = P["y_true"].astype(int)
    classes = [str(c) for c in P["class_names"]]; K = len(classes); N = len(y)
    model = HEASNet(K, pretrained=False); model.load_state_dict(torch.load(f"{R}/best_HEASNet.pt", map_location="cpu", weights_only=True)); model.to(DEV).eval()
    base = predict(model, paths).argmax(1); bok = base == y
    print(f"unmasked: {bok.sum()}/{N} correct = {100*bok.mean():.2f}% | identical to saved predictions: {(base == P['y_pred']).sum()}/{N}")
    # masks: same detection, same control generator and order as v2 -> identical masks
    rng = np.random.default_rng(SEED); G, Tx, CG, CT, BB = [], [], [], [], []
    for i, p in enumerate(paths):
        g, t, f = detect_overlays(p); cg, _ = control_mask(g, g | t, f, rng); ct, _ = control_mask(t, g | t | cg, f, rng)
        G.append(pack(g)); Tx.append(pack(t)); CG.append(pack(cg)); CT.append(pack(ct)); BB.append(black_corner_box(p))
        if (i + 1) % 500 == 0: print(f"  masks {i+1}/{N}")
    BB = np.array(BB); hasG = np.array([unpack(m).any() for m in G])
    pres = pd.DataFrame({"class": [disp(c) for c in classes], "n_test": np.bincount(y, minlength=K),
                         "green_box_%": [100 * hasG[y == c].mean() for c in range(K)], "black_corner_box_%": [100 * BB[y == c].mean() for c in range(K)]})
    pres.to_csv(f"{od}/{dt}_A6b_corner_box_presence.csv", index=False); print("\nPresence of a green box and of a pre-existing BLACK box in the bottom-left corner, per class"); print(pres.round(1).to_string(index=False))
    both = [pack(unpack(a) | unpack(b)) for a, b in zip(G, Tx)]
    conds = {"green box": G, "border text/graphics": Tx, "both overlays": both, "control: green-box shape moved": CG, "control: text area moved": CT}
    rows, trans, allpred = [], [], {"unmasked": base}
    for fill in ["black", "inpaint"]:
        for name, M in conds.items():
            pr = predict(model, paths, M, fill).argmax(1); ok = pr == y; allpred[f"{fill} | {name}"] = pr
            has = np.array([unpack(m).any() for m in M]); a = int((bok & ~ok).sum()); b = int((~bok & ok).sum()); cl = pd.Series(ok).groupby(y).mean() - pd.Series(bok).groupby(y).mean()
            ch = pr != base; to_pol = int((ch & (pr == classes.index("polyps"))).sum()) if "polyps" in classes else np.nan
            rows.append({"fill": fill, "masked_region": name, "images_with_region": int(has.sum()), "accuracy_masked_%": 100 * ok.mean(), "change_pp_all": 100 * (ok.mean() - bok.mean()),
                         "change_pp_images_with_region": 100 * (ok[has].mean() - bok[has].mean()) if has.any() else 0.0, "predictions_changed": int(ch.sum()),
                         "changed_to_polyps": to_pol, "correct_to_wrong": a, "wrong_to_correct": b, "exact_McNemar_p": float(binomtest(a, a + b, .5).pvalue) if a + b else 1.0,
                         "largest_class_drop": f"{disp(classes[int(cl.idxmin())])} ({100*cl.min():+.1f} pp)"})
            for c in range(K):
                for d in range(K):
                    n = int(((y == c) & ch & (pr == d)).sum())
                    if n: trans.append({"fill": fill, "masked_region": name, "true_class": disp(classes[c]), "new_prediction": disp(classes[d]), "images": n, "%_of_true_class": 100 * n / (y == c).sum()})
            print(f"  {fill:7s} | {name:32s} | acc {100*ok.mean():6.2f}% ({100*(ok.mean()-bok.mean()):+6.2f} pp) | changed {ch.sum():4d} | -> polyps {to_pol}")
    d = pd.DataFrame(rows); d.to_csv(f"{od}/{dt}_A6b_overlay_masking_black_vs_inpaint.csv", index=False)
    print(f"\n{ds}: unmasked accuracy {100*bok.mean():.2f}%"); print(d.round(3).to_string(index=False))
    tr = pd.DataFrame(trans, columns=["fill", "masked_region", "true_class", "new_prediction", "images", "%_of_true_class"]).sort_values(["fill", "masked_region", "images"], ascending=[True, True, False]); tr.to_csv(f"{od}/{dt}_A6b_prediction_transitions.csv", index=False)
    print("\nWhere the changed predictions go (top 8 per fill, condition 'green box' and 'both overlays')")
    for fill in ["black", "inpaint"]:
        for name in ["green box", "both overlays"]:
            print(f"--- {fill} | {name}"); print(tr[(tr.fill == fill) & (tr.masked_region == name)].head(8).round(1).to_string(index=False))
    pc = pd.DataFrame({"class": [disp(c) for c in classes], "accuracy_unmasked_%": [100 * bok[y == c].mean() for c in range(K)],
                       **{f"{k} change_pp": [100 * ((v == y)[y == c].mean() - bok[y == c].mean()) for c in range(K)] for k, v in allpred.items() if k != "unmasked"}})
    pc.to_csv(f"{od}/{dt}_A6b_per_class_change.csv", index=False)
    np.savez_compressed(f"{od}/{dt}_A6b_predictions_all_conditions.npz", paths=np.asarray(paths), y_true=y, class_names=np.asarray(classes), black_corner_box=BB, **{re.sub(r"[^A-Za-z0-9]+", "_", k): v for k, v in allpred.items()})
    # figure: same image, both fills (one image per class with a green box, else any)
    ex = []
    for c in range(K):
        idx = np.where(y == c)[0]; w = idx[hasG[idx]]; ex.append(int((w if len(w) else idx)[0]))
    fig, ax = plt.subplots(len(ex), 3, figsize=(8.4, 2.8 * len(ex)), squeeze=False)
    for r, i in enumerate(ex):
        m = unpack(both[i]); ax[r, 0].imshow(make_input(paths[i])[1]); ax[r, 1].imshow(make_input(paths[i], m, "black")[1]); ax[r, 2].imshow(make_input(paths[i], m, "inpaint")[1])
        ax[r, 0].set_ylabel(disp(classes[y[i]]), fontsize=9, fontweight="bold")
    for k, t in enumerate(["Input", "Both overlays masked: black fill", "Both overlays masked: inpainted fill"]): ax[0, k].set_title(t, fontsize=8)
    for a in ax.ravel(): a.set_xticks([]); a.set_yticks([])
    fig.suptitle(f"{ds}: overlay masking with black and inpainted fill (exact model inputs)", y=1.0); fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_A6b_black_vs_inpaint.png")
    fig, axx = plt.subplots(figsize=(max(6, .7 * K), 3.8)); xk = np.arange(K); wd = .27
    for j, (col, lb, cc) in enumerate([("black | both overlays change_pp", "both overlays, black fill", "#c0392b"), ("inpaint | both overlays change_pp", "both overlays, inpainted fill", "#2471a3"),
                                       ("black | control: green-box shape moved change_pp", "control, black fill", "#7f8c8d")]):
        axx.bar(xk + (j - 1) * wd, pc[col], wd, label=lb, color=cc)
    axx.axhline(0, c="k", lw=.8); axx.set_xticks(xk); axx.set_xticklabels(pc["class"], rotation=45, ha="right"); axx.set_ylabel("Change in class recall (pp)"); axx.legend(fontsize=8)
    axx.set_title(f"{ds}: effect of masking the overlays per class (black vs inpainted fill)"); fig.tight_layout(); save(fig, f"{od}/{dt}_Fig_A6b_per_class_change.png")
    del model; gc.collect(); torch.cuda.empty_cache() if torch.cuda.is_available() else None

for ds, sub in [("HyperKvasir", "HyperKvasir"), ("Kvasir-V2", "Kvasir_V2")]:
    try: run(ds, sub)
    except Exception as e:
        import traceback; traceback.print_exc(); print(f"!! {ds} FAILED: {e}")
banner(f"ALL DONE in {(time.time()-T0)/60:.1f} min"); _LOG.flush()
zp = OUT + ".zip"
with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
    for r, _, fs in os.walk(OUT):
        for f in fs: z.write(os.path.join(r, f), os.path.relpath(os.path.join(r, f), os.path.dirname(OUT)))
print(f"ZIP: {zp} ({os.path.getsize(zp)/1e6:.1f} MB)")
try:
    from IPython.display import FileLink, display; display(FileLink(os.path.relpath(zp)))
except Exception: pass
```
